```python
import pandas as pd
import numpy as np

from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)

import matplotlib.pyplot as plt
```

```python
from scipy import sparse
```

```python
X_train = sparse.load_npz(
    "../../artifacts/X_train_tree.npz"
)

X_test = sparse.load_npz(
    "../../artifacts/X_test_tree.npz"
)
```

```python
import pandas as pd
```

```python
y_train = pd.read_csv(
    "../../artifacts/y_train.csv"
)

y_test = pd.read_csv(
    "../../artifacts/y_test.csv"
)
```

```python
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)

print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
y_train = y_train.squeeze()
y_test = y_test.squeeze()
```

```python
print(y_train.shape)
print(y_test.shape)
```

```python
print(y_train.value_counts())
```

```python
from sklearn.ensemble import RandomForestClassifier


rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    class_weight="balanced",
    n_jobs=-1
)
```

```python
rf_model.fit(
    X_train,
    y_train
)
```

```python
y_pred = rf_model.predict(X_test)
```

```python
y_prob = rf_model.predict_proba(X_test)
```

```python
y_prob.shape
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)
```

```python
accuracy = accuracy_score(
    y_test,
    y_pred
)

print("Accuracy:", accuracy)
```

```python
precision = precision_score(
    y_test,
    y_pred,
    average="weighted"
)

print("Precision:", precision)
```

```python
recall = recall_score(
    y_test,
    y_pred,
    average="weighted"
)

print("Recall:", recall)
```

```python
f1 = f1_score(
    y_test,
    y_pred,
    average="weighted"
)

print("F1 Score:", f1)
```

```python
print(
    classification_report(
        y_test,
        y_pred
    )
)
```

```python
cm = confusion_matrix(
    y_test,
    y_pred
)

print(cm)
```

```python
import pandas as pd


feature_names = pd.read_csv(
    "../../artifacts/feature_names_tree.csv"
)


importance = pd.DataFrame({
    "Feature": feature_names.iloc[:,0],
    "Importance": rf_model.feature_importances_
})


importance = importance.sort_values(
    by="Importance",
    ascending=False
)


importance.head(20)
```

```python
from sklearn.model_selection import StratifiedKFold, cross_val_score
```

```python
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)
```

```python
cv_scores = cross_val_score(
    rf_model,
    X_train,
    y_train,
    cv=cv,
    scoring="f1_weighted",
    n_jobs=-1
)
```

```python
print("F1 Scores for each fold:")
print(cv_scores)

print("\nAverage F1 Score:")
print(cv_scores.mean())

print("\nStandard Deviation:")
print(cv_scores.std())
```
