```python
from nltk.corpus import stopwords
```

```python
stop_words = set(stopwords.words("english"))
```

```python
print(list(stop_words)[:10])
```

```python
def clean_text(text):

    text = text.lower()

    text = re.sub(
        r"[^a-z\s]",
        "",
        text
    )

    words = text.split()

    words = [
        word 
        for word in words 
        if word not in stop_words
    ]

    return words
```

```python
text_df["tokens"] = (
    text_df["VIOLATION_DESCRIPTION"]
    .apply(clean_text)
)
```
